```r
library(tidyverse)
library(data.table)
library(cowplot)
library(ggrepel)
```

```r
sample_description <- '/faststorage/project/primatediversity/data/gVCFs_recalling_10_12_2024_metadata/plots/SupTable_Sample_Stats_wGT_QC_MQ.tsv'
sample_description_df <- fread(sample_description)
```

```r
sample_description_df %>%
names()
```

```r
sample_description_df <- sample_description_df %>%

mutate(f_mapped = cut(f_mapped_reads, c(0, 0.5, 0.75, 0.9, 0.95, 1), include.lowest = T)) 
```

```r
S2A <- sample_description_df %>%

ggplot()+
geom_histogram(aes(x = MQ1kb, fill = f_mapped), bins = 61)+
scale_fill_viridis_d(direction = -1)+

scale_x_continuous(breaks = seq(0, 60, 5))+
xlab('Genome-average mapping quality')+
ylab('N of samples')+
labs(fill = "Fraction of\nmapped reads")+
        theme_bw() + 
        theme(
            text = element_text(size = 18), 
            axis.title.x = element_blank(),
            axis.text.x = element_blank(),
            axis.ticks.x = element_blank()
        )
```

```r
S2B <- sample_description_df %>%

ggplot()+
# geom_point(aes(x = f_mapped_reads, y = MQ, color = div_chrA, shape = cov_bin), size = 3, alpha = 0.75, stroke = 2)+
geom_point(aes(y = f_mapped_reads, x = MQ, fill = div_chrA), shape = 21, size = 4, alpha = 0.5)+
# scale_color_viridis_d(direction = 1)+
scale_fill_viridis_c(direction = 1, option = 2)+


scale_x_continuous(breaks = seq(0, 60, 5))+
xlab('Genome-average mapping quality')+
ylab('Fraction of mapped reads')+
# scale_x_log10()+
# scale_y_log10()+
facet_wrap(~cov_bin, nrow = 4)+
labs(fill = "Divergence\nfrom\nreference")+
        theme_bw() + 
        theme(
            text = element_text(size = 18)
        )
```

```r
figure <- plot_grid(S2A, S2B, labels = c('A', 'B'), label_size = 16, nrow = 2, align = 'v', rel_heights = c(1.25, 4))
```
